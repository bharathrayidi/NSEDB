# Interactive SQL Query Widget with Foundry‑style Display

This notebook lets you query a table, optionally filter by symbol, and view results using the unified `display_dataframe_full` function which provides a Foundry‑style interactive preview and a *Show Statistics* button.

````python
import sqlite3
import pandas as pd
import ipywidgets as widgets
from IPython.display import display
from display import foundry_pro_display

# Path to the SQLite database (adjust if needed)
db_path = r"d:/badri/vs code/NSEDB-main/market_data.db"
conn = sqlite3.connect(db_path)
````

````python
# Table selector
table_widget = widgets.Dropdown(
    options=[
        "cash_market_full",
        "index_derivative",
        "stocks_derivative",
        "nifty_predictions",
        "nifty_daily_forecast",
        "nifty_forecast_3month",
        "nifty_model_performance",
    ],
    description="Table:",
    value="cash_market_full"
)
````

````python
# Run query button
run_button = widgets.Button(description="Run Query")

def run_query(_):
    table = table_widget.value
    # symbol = symbol_widget.value
    # where_clause = f"WHERE Symbol = '{symbol}'" if symbol else ""
    query = f"""
        SELECT *
        FROM {table}
        ORDER BY Date DESC
    """
    df = pd.read_sql_query(query, conn)
    # Use the unified Foundry‑style display function
    foundry_pro_display(df)

run_button.on_click(run_query)
````

````python
# Show widgets together
display(table_widget, run_button)
````
